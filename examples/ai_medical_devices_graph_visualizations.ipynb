
## Разбор `mindmap`

Мини-парсер ниже предназначен для отступного синтаксиса Mermaid `mindmap`: два пробела означают один уровень. Каждому вхождению присваивается собственный числовой ID, поэтому одинаковые подписи компаний не склеиваются. `<br/>` преобразуется в перенос строки.


```python

from collections import defaultdict
import re
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D


def parse_mermaid_mindmap(source):
    """Return labels, parent map, children and depths from Mermaid mindmap text."""
    rows = [line.rstrip() for line in source.splitlines() if line.strip()]
    if not rows or rows[0].strip() != "mindmap":
        raise ValueError("Ожидается Mermaid-диаграмма, начинающаяся с 'mindmap'")

    labels, parent, children, depth = {}, {}, defaultdict(list), {}
    stack = {}
    for node_id, line in enumerate(rows[1:]):
        spaces = len(line) - len(line.lstrip(" "))
        if spaces % 2:
            raise ValueError(f"Отступ должен быть кратен двум: {line!r}")
        level = spaces // 2 - 1
        if level < 0:
            raise ValueError(f"Некорректный отступ: {line!r}")
        text = line.strip()
        if level == 0:
            text = re.sub(r"^root\(\((.*)\)\)$", lambda match: match.group(1), text)
        text = re.sub(r"<br\s*/?>", "\n", text, flags=re.I)
        labels[node_id], depth[node_id], stack[level] = text, level, node_id
        if level:
            if level - 1 not in stack:
                raise ValueError(f"Нет родителя для: {text}")
            parent[node_id] = stack[level - 1]
            children[parent[node_id]].append(node_id)
        for stale in [key for key in stack if key > level]:
            del stack[stale]

    roots = [node for node in labels if node not in parent]
    if len(roots) != 1:
        raise ValueError(f"Ожидался один корень, получено: {len(roots)}")
    return labels, parent, children, depth, roots[0]


labels, parent, children, depth, root = parse_mermaid_mindmap(mermaid_code)
edges = list(parent.items())  # (child, parent)
leaves = [node for node in labels if not children[node]]
print(f"Вершин: {len(labels)}; рёбер: {len(edges)}; изделий-листьев: {len(leaves)}; глубина: {max(depth.values())}")
assert len(edges) == len(labels) - 1
assert len(leaves) == 59, "Заголовок сообщает о 59 изделиях — структура должна совпадать"
```


## 1. Обычная визуализация

Это детерминированная слева-направо раскладка дерева: листья равномерно размещаются по вертикали, а родитель — в центре диапазона своих детей. Цвет показывает уровень иерархии. При необходимости сохраните рисунок, раскомментировав `savefig`.


```python

def tidy_tree_positions(root, children, depth):
    """Place leaves on consecutive rows and parents at their children's barycentre."""
    positions, cursor = {}, [0]
    def visit(node):
        if not children[node]:
            y = cursor[0]
            cursor[0] += 1
        else:
            child_y = [visit(child) for child in children[node]]
            y = float(np.mean(child_y))
        positions[node] = (depth[node], y)
        return y
    visit(root)
    return positions


palette = {0: "#7b2cbf", 1: "#1976d2", 2: "#00a896", 3: "#f4a261"}
pos = tidy_tree_positions(root, children, depth)
fig, ax = plt.subplots(figsize=(30, 22))

for child, par in edges:
    x0, y0 = pos[par]
    x1, y1 = pos[child]
    elbow = x0 + 0.45 * (x1 - x0)
    ax.plot([x0, elbow, elbow, x1], [y0, y0, y1, y1],
            color="#a8b3c1", linewidth=0.75, zorder=1)

for node, (x, y) in pos.items():
    color = palette[depth[node]]
    size = {0: 300, 1: 160, 2: 90, 3: 30}[depth[node]]
    ax.scatter(x, y, s=size, color=color, edgecolor="white", linewidth=0.7, zorder=2)
    if depth[node] < 3:
        ax.text(x + 0.035, y, labels[node], va="center", ha="left", fontsize=8.5,
                weight="bold" if depth[node] <= 1 else "normal",
                bbox=dict(boxstyle="round,pad=.2", fc="white", ec="none", alpha=.82))
    else:
        ax.text(x + 0.035, y, labels[node].replace("\n", " · "), va="center", ha="left",
                fontsize=6.2, color="#263238")

ax.set_title("AI-медизделия — обычная иерархическая визуализация", fontsize=20, pad=20)
ax.set_xlim(-0.2, max(depth.values()) + 2.25)
ax.set_ylim(len(leaves) + 1, -2)
ax.axis("off")
plt.tight_layout()
# fig.savefig("ai_medical_devices_ordinary.png", dpi=180, bbox_inches="tight")
plt.show()
```


## 2. Гиперболическая визуализация в диске Пуанкаре

Для кривизны $K=-1$ вершина глубины $d$ получает евклидов радиус
$r_d=\tanh(\alpha d)$ — равные шаги в касательном пространстве становятся всё более тесными у границы. Листьям выделяются равные угловые интервалы, внутренние вершины получают круговое среднее углов детей.

Рёбра — **не прямые хорды**, а геодезики Пуанкаре, вычисленные через сложение Мёбиуса. Таким образом, дуги корректны для геометрии используемой модели.


```python

def mobius_add(u, v, c=1.0):
    """Möbius addition for Euclidean coordinates inside the Poincaré ball."""
    u, v = np.asarray(u, float), np.asarray(v, float)
    uv, u2, v2 = u @ v, u @ u, v @ v
    denominator = 1 + 2*c*uv + c*c*u2*v2
    return ((1 + 2*c*uv + c*v2)*u + (1 - c*u2)*v) / denominator


def mobius_scalar_mul(t, x, c=1.0):
    """Möbius scalar multiplication inside the Poincaré ball."""
    x = np.asarray(x, float)
    norm = np.linalg.norm(x)
    if norm == 0:
        return np.zeros_like(x)
    z = np.clip(np.sqrt(c)*norm, 0, 1 - 1e-12)
    return np.tanh(t*np.arctanh(z))*x/(np.sqrt(c)*norm)


def geodesic(u, v, c=1.0, samples=36):
    """Sample the Poincaré geodesic joining u and v."""
    relative = mobius_add(-np.asarray(u), v, c)
    return np.array([mobius_add(u, mobius_scalar_mul(t, relative, c), c)
                     for t in np.linspace(0, 1, samples)])


def angular_layout(root, children, depth, radial_step=0.42):
    """Allocate equal angular space to leaves and map levels into the disk."""
    angle, leaf_cursor = {}, [0]
    leaf_count = sum(not children[node] for node in labels)
    def visit(node):
        if not children[node]:
            theta = 2*np.pi*(leaf_cursor[0] + 0.5)/leaf_count
            leaf_cursor[0] += 1
        else:
            child_angles = np.array([visit(child) for child in children[node]])
            theta = np.angle(np.mean(np.exp(1j*child_angles))) % (2*np.pi)
        angle[node] = theta
        return theta
    visit(root)
    points = {}
    for node in labels:
        radius = np.tanh(radial_step * depth[node])
        points[node] = radius*np.array([np.cos(angle[node]), np.sin(angle[node])])
    points[root] = np.zeros(2)
    return points, angle


points, angles = angular_layout(root, children, depth)
assert all(np.linalg.norm(point) < 1 for point in points.values())

fig, ax = plt.subplots(figsize=(20, 20))
ax.add_patch(plt.Circle((0, 0), 1, facecolor="#f8fbff", edgecolor="#263238",
                        linewidth=2.2, zorder=0))

for child, par in edges:
    curve = geodesic(points[par], points[child])
    ax.plot(curve[:, 0], curve[:, 1], color="#90a4ae", linewidth=.75,
            alpha=.82, zorder=1)

for level in sorted(set(depth.values())):
    nodes = [node for node in labels if depth[node] == level]
    xy = np.array([points[node] for node in nodes])
    ax.scatter(xy[:, 0], xy[:, 1], s={0:420, 1:170, 2:75, 3:24}[level],
               color=palette[level], edgecolor="white", linewidth=.6, zorder=3,
               label=f"Уровень {level}")

# Internal labels are kept inside; leaf labels are rotated around the rim.
for node, point in points.items():
    if depth[node] == 0:
        ax.text(*point, "AI\n59", ha="center", va="center", color="white",
                fontsize=10, weight="bold", zorder=4)
    elif depth[node] <= 2:
        ax.text(*(point*1.025), labels[node], ha="center", va="center", fontsize=6.4,
                bbox=dict(boxstyle="round,pad=.16", fc="white", ec=palette[depth[node]], alpha=.88),
                zorder=4)
    else:
        theta = angles[node]
        anchor = 1.025*np.array([np.cos(theta), np.sin(theta)])
        degrees = np.degrees(theta)
        right_half = np.cos(theta) >= 0
        rotation = degrees if right_half else degrees + 180
        # Product name (middle Mermaid line) is the most useful compact leaf label.
        parts = labels[node].split("\n")
        short_label = parts[1] if len(parts) >= 2 else parts[0]
        ax.text(*anchor, short_label, rotation=rotation, rotation_mode="anchor",
                ha="left" if right_half else "right", va="center", fontsize=5.5,
                color="#263238", clip_on=False, zorder=4)

ax.set(xlim=(-1.28, 1.28), ylim=(-1.28, 1.28), aspect="equal")
ax.set_title("AI-медизделия в модели диска Пуанкаре ($K=-1$)", fontsize=20, pad=22)
ax.axis("off")
ax.legend(loc="lower center", ncol=4, frameon=False, bbox_to_anchor=(.5, -.035))
plt.tight_layout()
# fig.savefig("ai_medical_devices_poincare.png", dpi=200, bbox_inches="tight")
plt.show()
```


## Как адаптировать notebook

- Замените только содержимое `mermaid_code`, сохранив синтаксис `mindmap` и отступ в два пробела на уровень.
- Для более сильного сжатия к границе увеличьте `radial_step` в `angular_layout`.
- Для подписей всех полей изделия замените `short_label` на `labels[node].replace("\n", " · ")`.
- Раскомментируйте `fig.savefig(...)` в нужной ячейке для экспорта PNG.
